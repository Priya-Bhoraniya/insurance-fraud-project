```python
import numpy as np # for a calculation
import matplotlib.pyplot as plt # for a graph
```

```python
X = df['vehicle_price'].values.astype(float)
Y = df['annual premium'].values.astype(float) # array ma convert karva mate
```

```python
#variable insilize.
# We initialize the variables before starting the algorithm. The values of m and c are updated in every iteration, and the loss of each epoch is stored in loss_history.

n = len(X) # n means total  number of data points 

m = 0 # slope
c = 0. #Intercept.

lr = 0.000000001
epochs = 10000. #Epoch means ketli var dataset  par training karvi.

loss_history = [] # badha epoch no lose history store karva.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# X and Y
X = df['vehicle_price'].values.astype(float)
Y = df['annual premium'].values.astype(float)

# Initialize variables
n = len(X)
m = 0.0
c = 0.0

learning_rate = 0.0000000001   # 1e-10
epochs = 10000

loss_history = []

# Gradient Descent
for i in range(epochs):

    # Prediction the value
    Y_pred = m * X + c

    # Error
    error = Y_pred - Y #prediction ma error aave to e error find karva mate 

    # Loss (Mean Squared Error)
    loss = (1/n) * np.sum(error**2)
    loss_history.append(loss) # error ma ketlo loss thayo che e mate

    # Gradient
    dm = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)

    # Update parameters
    m = m - learning_rate * dm
    c = c - learning_rate * dc # pachi error remove karva mate m and c ni value update karva mate.

    # Print loss after every 1000 epochs
    if i % 1000 == 0:
        print(f"Epoch {i} : Loss = {loss:.2f}")

# Final values
print("\nFinal Slope (m):", m)
print("Final Intercept (c):", c)

# Prediction
vehicle_price = 30000

predicted_premium = m * vehicle_price + c

print("Predicted Annual Premium :", predicted_premium)

# Loss Graph
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss vs Epoch")
plt.show()

#  for a gradient decent        
        
```

```python
print("----- Comparison -----")

print("Scikit-Learn Slope :", model.coef_[0])
print("Gradient Descent Slope :", m)

print("Scikit-Learn Intercept :", model.intercept_)
print("Gradient Descent Intercept :", c)
```

```python
# Logistic Regression

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load dataset
df = pd.read_csv("insurance_fraud_data.csv")

# Remove rows where fraud reported is missing
df = df.dropna(subset=['fraud reported'])

# X = Input
X = df[['vehicle_price']]

# Y = Output
Y = df['fraud reported']

# Y માં N/Y છે, yes and no ne 0 & 1 ma convert karva mate
Y = Y.map({'N': 0, 'Y': 1})

# Split data into training and testing
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

# Create Logistic Regression model
model = LogisticRegression()

# Train model
model.fit(X_train, Y_train)

# Prediction
Y_pred = model.predict(X_test)

# Accuracy
accuracy = accuracy_score(Y_test, Y_pred)

print("Accuracy :", accuracy)

# Confusion Matrix ---- ---- કketla  fraud correctly detect thaya che and kyu model wrong che
print("\nConfusion Matrix:")
print(confusion_matrix(Y_test, Y_pred))

# Classification Report----precision,# recall,# f1-score# support
# prediction -- model ae fraud kryu aemathi ketla ae fraud karyo j to,
# recall---actual frauad mathi model ae ketla fraud sodhi kadya
# F1-score---predictiton and recall no combine
print("\nClassification Report:")
print(classification_report(Y_test, Y_pred))
```
